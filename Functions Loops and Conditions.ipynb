# The Automated Spam Filter

### 1. The Text Cleaner (Functions, Regex, & Strings)

```python
import re

def clean_email(text):
  updated_text = re.sub(r"[,.]", "", text)
  updated_text = updated_text.lower()
  words_list = updated_text.split()
  return words_list
```

### 2. The Threat Detector (Functions, Loops, Conditions, & Sets)

```python
def calculate_risk(cleaned_words, spam_words):
  risk_score = 0
  for word in cleaned_words:
    if word in spam_words:
      risk_score += 1
    
  return risk_score
```

### 3. Processing the Inbox (Dictionaries & Logic)

```python
spam_database = {'urgent', 'winner', 'free', 'action'}

incoming_email = {
    'sender': 'unknown@suspicious.com',
    'subject': 'URGENT Action Required',
    'body': 'You are a winner. Take action now for your free prize, action is required.'
    }

email_body = incoming_email['body']
cleaned_body = clean_email(email_body)
final_score = calculate_risk(cleaned_body, spam_database)

if final_score >= 2:
  print("Spam detected! Sending to junk.")
else:
  print("Email is safe. Sending to inbox.")
```

### 4. The Emergency Lockdown(While Loops)

```python
server_pings = [101, 102, 103, 999, 104, 105]

N = len(server_pings)
i = 0

while i < N:
  if server_pings[i] != 999:
    print("Connection {} is secure.".format(server_pings[i]))
  else:
    print("CRITICAL THREAT: IP 999 detected. Initiating Lockdown!")
    break
  i += 1
```
