### 1. The Location Standardizer(Functions, Strings, & Regex)

```python
import re
def clean_location(location):
  cleaned_text = re.sub(r"[.,*]", "", location)
  removed_spaces = cleaned_text.strip()
  final_upper_string = removed_spaces.upper()
  return final_upper_string
```
