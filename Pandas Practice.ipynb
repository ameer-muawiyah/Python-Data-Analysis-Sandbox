```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df.head()
```

```python
df['Survived'].value_counts(normalize=True) * 100
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.groupby('Sex')['Survived'].mean() * 100
```

```python
df.groupby(['Pclass', 'Sex'])['Survived'].mean() * 100
```

```python
rich_passengers = df[(df['Pclass'] == 1) & (df['Fare'] > 200)]
rich_passengers[['Name', 'Fare', 'Survived', 'Sex']].sort_values('Fare', ascending=False)
rich_passengers['Survived'].value_counts(normalize = True) * 100
```
