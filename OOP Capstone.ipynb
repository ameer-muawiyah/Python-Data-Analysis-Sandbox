# The Autonomous Drone Delivery

## 1. The Blueprint & The Setup

```python
import re
class DeliveryDrone:
  operating_company = "Global Cargo"

  def __init__ (self, drone_id, battery_level):
    self.drone_id = drone_id
    self.battery_level = battery_level
    self.is_grounded = False
    self.delivery_log = {}

# The Navigation System
  def clean_route(self, messy_data):

    cleaned_data = re.sub(r"[!-]", "", messy_data)
    
    upper_data = cleaned_data.upper()

    upper_data_list = upper_data.split()

    cleaned_routes = set(upper_data_list)

    return cleaned_routes

# The Flight Engine
  def execute_flight(self, messy_zones):

    cleaned_zones = self.clean_route(messy_zones)

    for zone in cleaned_zones:
      if self.battery_level >= 15:
        self.battery_level -= 15
        self.delivery_log[zone] = "Success"
      else:
        self.is_grounded = True
        self.delivery_log[zone] = "Failed"
        break
    print(self.delivery_log)

# Diagnostics
  def run_diagnostics(self):
    try:
      efficiency_score = len(self.delivery_log)/self.battery_level
      print("Efficiency score of Drone {} is {}".format(self.drone_id, efficiency_score))
    except Exception as e:
      print("Diagnostics failed due to system error: ", e)
```

## 2. Deployment

```python
droneAlpha = DeliveryDrone("Alpha", 50)
droneBeta = DeliveryDrone("Beta", 20)
DeliveryDrone.operating_company = "Apex Logistics"

droneAlpha.execute_flight("North-Zone! South-Zone! North-Zone! East-Zone-")
droneBeta.execute_flight("North-Zone! South-Zone! North-Zone! East-Zone-")

droneAlpha.run_diagnostics()
droneBeta.run_diagnostics()
```
