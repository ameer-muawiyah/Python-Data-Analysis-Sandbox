# The Nightly Server Backup (File Handling)

## 1. Generating the Daily Log(Writing)

```python
daily_events = [
    "Login: USER_001\n",

    "ERROR: Wrong Password\n",

    "Login: USER_002\n",

    "ERROR: Server Timeout\n",

    "Login: USER_003\n"
]

with open("daily_log.txt", 'w') as logFile:
  for event in daily_events:
    logFile.write(event)
```
