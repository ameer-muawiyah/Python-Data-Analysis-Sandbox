# The Nightly Server Backup (File Handling)

## 1. Generating the Daily Log(Writing)

```python
daily_events = [
    "Login: USER_001\n",

    "ERROR: Wrong Password\n",

    "Login: USER_002\n",

    "ERROR: Server Timeout\n",

    "Login: USER_003\n"
]

with open("daily_log.txt", 'w') as logFile:
  for event in daily_events:
    logFile.write(event)
```

## 2. The Migration (Reading and Appending)

```python
with open("daily_log.txt", 'r') as logFile, open("master_backup.txt", 'a') as backupFile:
  for line in logFile:
    if "Login" in line:
      backupFile.write(line)
```

## 3. The Reset (Destructive Overrite)

```python
with open("daily_log.txt", 'w') as logFile:
  logFile.write("Log cleared. Awaiting new data.")
```

```python
with open("master_backup.txt", 'r') as backupFile:
  for line in backupFile:
    print(line.strip())
```
