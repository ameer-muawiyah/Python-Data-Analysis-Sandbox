### 1. The Sensor Signal Decoder

```python
import re

def decode_signal(messy_signal):
  
  cleaned_signal = re.sub(r"[!@#]", "", messy_signal)

  strip_spaces = cleaned_signal.strip()
  
  decoded_signal = strip_spaces.lower()

  return decoded_signal
```

### 4. The Live Radar Feed

```python
live_radar = [
    ' @ASTEROID_X# ',
    ' COMET_Y! ',
    ' #SPACE_DEVRIS@ ',
    '!ASTEROID_Z!'
]

known_hazards = {
    'asteroid_x',
    'asteroid_z',
    'solar_flare'
}

evasion_counter = 0

for signature in live_radar:

  cleaned_signal = decode_signal(signature)

  if cleaned_signal in known_hazards:
    evasion_counter += 1
  else:
    print("Object {} is safe. Continuing on course.".format(cleaned_signal))
  
  if evasion_counter == 2:
    print("CRITICAL ALERT: Multiple hazards detected. Initiating Emergency Evasion maneuver!")
    break
if evasion_counter < 2:
  print("No critical hazards found. Everything is under control.")
```
