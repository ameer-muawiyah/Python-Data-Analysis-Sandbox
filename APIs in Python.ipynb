```python
import requests
import pandas as pd

# --- SYSTEM SETUP (DO NOT EDIT) ---
class MockResponse:
    status_code = 200
    def json(self):
        return [
            {"suspect_id": "V-011", "alias": "The Ghost", "threat_level": 89, "last_seen": "Downtown"},
            {"suspect_id": "V-012", "alias": "Cipher", "threat_level": 95, "last_seen": "Industrial District"},
            {"suspect_id": "V-014", "alias": "Apex", "threat_level": 72, "last_seen": "North Side"}
        ]

# Intercepting internet traffic for the simulation
requests.get = lambda url: MockResponse()
target_endpoint = "https://intel.gcpd.gov/api/v1/suspects"
# -----------------------------------
```

```python
response = requests.get(target_endpoint)

if response.status_code == 200:
  print("Connection Successful.")
  suspects_data = response.json()
  print(suspects_data)
else:
  print("Critical Failure. Status code returned is:", response.status_code)
```
