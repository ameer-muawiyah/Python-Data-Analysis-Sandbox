## Customer Tags (Sets)

```python
premium_tags = {'electronics', 'VIP', 'fast_shipping'}
standard_tags = {'books', 'electronics', 'standard_shipping'}
premium_tags
standard_tags
```

```python
premium_tags.add('loyalty_member')
premium_tags
```

```python
premium_tags.remove('fast_shipping')
premium_tags
```

```python
common_tags = premium_tags & standard_tags
common_tags
```

```python
list_of_tags = premium_tags | standard_tags
list_of_tags
```

```python
only_premium_tags = premium_tags - standard_tags
only_premium_tags
```
