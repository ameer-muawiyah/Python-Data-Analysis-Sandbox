## Part 1: Managing the Patient Queue (Lists)

```python
patient_queue = ["Emma", "John", "Sarah", "John"]
```

```python
print(patient_queue[0])
```

```python
patient_queue[2] = "Michael"
patient_queue
```

```python
patient_queue.append("David")
patient_queue
```

```python
patient_queue.insert(1, "Sophia")
patient_queue
```

```python
patient_queue.extend(["Olivia", "Liam"])
patient_queue
```

```python
patient_queue.count("John")
```

```python
patient_queue.remove("John")
patient_queue
```

```python
discharge_Liam = patient_queue.pop()
print("{} has been discharged".format(discharge_Liam))
```

```python
del patient_queue[2]
patient_queue
```

```python
queue_backup = patient_queue[:]
queue_backup
```

```python
patient_queue.sort()
patient_queue
```

```python
patient_queue.sort(reverse=True)
patient_queue
```

```python
print(patient_queue[0:3])
```

## Analyzing Vitals (Tuples & Basic Data Types)

```python
heart_rates = (72, 85, 90.5, 85, 76, 110, 85)
heart_rates
```

```python
len(heart_rates)
```

```python
highest_heart_rate = max(heart_rates)
highest_heart_rate
```

```python
lowest_heart_rate = min(heart_rates)
lowest_heart_rate
```

```python
sum_heart_rates = sum(heart_rates)
sum_heart_rates
```

```python
heart_rates.count(85)
```

```python
heart_rates.index(110)
```

```python
is_high_risk = True
print(f"Patient highest heart rate was {highest_heart_rate}, and high risk status is {is_high_risk}")
```
