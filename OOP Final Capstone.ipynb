# The Airline Flight Manifest System

## The Blueprint

```python
import re
class Flight:
  airline_name = "Skyward Airlines"
  
# CONSTRUCTOR
  def __init__(self, flight_num):
    self.flight_num = flight_num
    self.seating_inventory = {"First Class": 1, "Economy Class": 3}
    self.boarded_passengers = {}
    self.rejected_passengers = {}

# DATA CLEANER  
  def dataCleaner(self, messy_queue):
    cleaned_ids = re.sub(r"[!*#.-]", "", messy_queue)
    cleaned_queue = cleaned_ids.upper().split()
    return cleaned_queue

# SECURITY CLEARENCE
  def securityClearence(self, no_fly, vip):
    self.flagged_passengers = no_fly | vip

    self.banned_vips = no_fly & vip

    self.safe_vip = vip - self.banned_vips

# BOARDING ENGINE
  def boardingEngine(self, passenger_records, restricted_list, vip_list):

    total_cleaned_passengers = self.dataCleaner(passenger_records)
    self.securityClearence(restricted_list, vip_list)

    for passenger in total_cleaned_passengers:
      if passenger in restricted_list:
        print("passenger {} has been flagged restricted.\n".format(passenger))
        self.rejected_passengers[passenger] = "Rejected"
      elif passenger in self.safe_vip:
        if self.seating_inventory["First Class"]:
          print("Passenger {} has been assigned to First Class.\n".format(passenger))
          self.seating_inventory["First Class"] -= 1
          self.boarded_passengers[passenger] = "First Class"
        else:
          if self.seating_inventory["Economy Class"]:
            print("No First Class seats remaining. Passenger {} has been downgraded to Economy.\n".format(passenger))
            self.seating_inventory["Economy Class"] -= 1
            self.boarded_passengers[passenger] = "Economy Class"
          else:
            print("The Flight is full. Passenger {} cannot fit in.\n".format(passenger))
            self.rejected_passengers[passenger] = "Failed to Board"
      else:
        if self.seating_inventory["Economy Class"]:
          print("Standard passenger! Seat Assigned to passenger {}.\n".format(passenger))
          self.seating_inventory["Economy Class"] -= 1
          self.boarded_passengers[passenger] = "Economy Class"
        else:
          print("All seats are filled. No empty seat available for passenger {}.\n".format(passenger))
          self.rejected_passengers[passenger] = "Failed to Board"
        
# AUDIT AND DIAGNOSTICS        
  def diagnostics(self, faa_data):
    self.max_flight_time = max(faa_data)
    self.min_flight_time = min(faa_data)

    try:
      self.boarding_efficiency_score = len(self.boarded_passengers)/len(self.rejected_passengers)
    except Exception as e:
      print("Can't find the Efficiency Score: ", e)
```
